```python
import folium
from folium.plugins import HeatMap
import pandas as pd
import ipywidgets as widgets
from ipywidgets import interact, interact_manual

# Verifique se as colunas 'latitude' e 'longitude' são strings antes de aplicar .str.replace
if df_copia['latitude'].dtype == 'object':
    df_copia['latitude'] = df_copia['latitude'].str.replace(',', '.').astype(float)
if df_copia['longitude'].dtype == 'object':
    df_copia['longitude'] = df_copia['longitude'].str.replace(',', '.').astype(float)

# Remover valores nulos nas colunas de latitude e longitude
df_copia = df_copia.dropna(subset=['latitude', 'longitude'])

# Lista das features para as caixas de seleção
features = ['br', 'idade', 'dia_semana', 'horario', 'uf', 'municipio', 'fase_dia', 'sentido_via', 'condicao_metereologica', 'tipo_pista', 'tracado_via', 'uso_solo', 'sexo']

# Função para criar o mapa com base nas seleções
def criar_mapa(br=None, idade=None, dia_semana=None, horario=None, uf=None, municipio=None, fase_dia=None, sentido_via=None, condicao_metereologica=None, tipo_pista=None, tracado_via=None, uso_solo=None, sexo=None):
    # Filtrar os dados com base nas seleções
    df_filtrado = df_copia.copy()
    
    if br is not None:
        df_filtrado = df_filtrado[df_filtrado['br'] == br]
    if idade is not None:
        df_filtrado = df_filtrado[df_filtrado['idade'] == idade]
    if dia_semana is not None:
        df_filtrado = df_filtrado[df_filtrado['dia_semana'] == dia_semana]
    if horario is not None:
        df_filtrado = df_filtrado[df_filtrado['horario'] == horario]
    if uf is not None:
        df_filtrado = df_filtrado[df_filtrado['uf'] == uf]
    if municipio is not None:
        df_filtrado = df_filtrado[df_filtrado['municipio'] == municipio]
    if fase_dia is not None:
        df_filtrado = df_filtrado[df_filtrado['fase_dia'] == fase_dia]
    if sentido_via is not None:
        df_filtrado = df_filtrado[df_filtrado['sentido_via'] == sentido_via]
    if condicao_metereologica is not None:
        df_filtrado = df_filtrado[df_filtrado['condicao_metereologica'] == condicao_metereologica]
    if tipo_pista is not None:
        df_filtrado = df_filtrado[df_filtrado['tipo_pista'] == tipo_pista]
    if tracado_via is not None:
        df_filtrado = df_filtrado[df_filtrado['tracado_via'] == tracado_via]
    if uso_solo is not None:
        df_filtrado = df_filtrado[df_filtrado['uso_solo'] == uso_solo]
    if sexo is not None:
        df_filtrado = df_filtrado[df_filtrado['sexo'] == sexo]
    
    # Criar o mapa
    mapa = folium.Map(location=[-15.77972, -47.92972], zoom_start=5)
    
    # Adicionar o mapa de calor
    HeatMap(list(zip(df_filtrado['latitude'], df_filtrado['longitude']))).add_to(mapa)
    
    # Exibir o mapa
    return mapa

# Criar interatividade com caixas de seleção para as features
interact_manual(
    criar_mapa,
    br=widgets.Dropdown(options=[None] + list(df_copia['br'].unique()), description='BR:'),
    idade=widgets.Dropdown(options=[None] + list(df_copia['idade'].unique()), description='Idade:'),
    dia_semana=widgets.Dropdown(options=[None] + list(df_copia['dia_semana'].unique()), description='Dia da Semana:'),
    horario=widgets.Dropdown(options=[None] + list(df_copia['horario'].unique()), description='Horário:'),
    uf=widgets.Dropdown(options=[None] + list(df_copia['uf'].unique()), description='UF:'),
    municipio=widgets.Dropdown(options=[None] + list(df_copia['municipio'].unique()), description='Município:'),
    fase_dia=widgets.Dropdown(options=[None] + list(df_copia['fase_dia'].unique()), description='Fase do Dia:'),
    sentido_via=widgets.Dropdown(options=[None] + list(df_copia['sentido_via'].unique()), description='Sentido da Via:'),
    condicao_metereologica=widgets.Dropdown(options=[None] + list(df_copia['condicao_metereologica'].unique()), description='Condição Metereológica:'),
    tipo_pista=widgets.Dropdown(options=[None] + list(df_copia['tipo_pista'].unique()), description='Tipo de Pista:'),
    tracado_via=widgets.Dropdown(options=[None] + list(df_copia['tracado_via'].unique()), description='Traçado da Via:'),
    uso_solo=widgets.Dropdown(options=[None] + list(df_copia['uso_solo'].unique()), description='Uso do Solo:'),
    sexo=widgets.Dropdown(options=[None] + list(df_copia['sexo'].unique()), description='Sexo:')
)
```
